[**第 1 章：g-h 滤波（The g-h Filter）**](01_g-h_filter.html)

直观介绍 g-h 滤波（g-h filter），该滤波器族包含卡尔曼滤波（Kalman filter）。本章绝非凑数——读完本章，你将理解卡尔曼滤波背后的核心概念。


[**第 2 章：离散贝叶斯滤波（The Discrete Bayes Filter）**](02_Discrete_Bayes.html)

介绍离散贝叶斯滤波（discrete Bayes filter）。你将由此学习支撑卡尔曼滤波的、易于理解的概率（贝叶斯，Bayesian）推理。


[**第 3 章：最小二乘滤波（Least Squares Filter）**](03_Least_Squares_Filters.html)

介绍批量形式与递推形式的最小二乘滤波（least squares filter）。该章内容我尚未开始撰写。


[**第 4 章：高斯概率（Gaussian Probabilities）**](04_Gaussians.html)

介绍在贝叶斯（Bayesian）意义下用高斯分布（Gaussian）表示信念。高斯分布使离散贝叶斯滤波中的算法可应用于连续域。


[**第 5 章：一维卡尔曼滤波（One Dimensional Kalman Filters）**](05_Kalman_Filters.html)

通过将离散贝叶斯滤波（discrete Bayesian filter）改为使用高斯分布，实现卡尔曼滤波（Kalman filter）。这是功能完整的一维卡尔曼滤波，但仅适用于一维问题。


[**第 6 章：多元卡尔曼滤波（Multivariate Kalman Filter）**](/06_Multivariate_Kalman_Filters.html)

将上一章的卡尔曼滤波推广为完整的广义滤波器。


[**第 7 章：卡尔曼滤波数学（Kalman Filter Math）**](07_Kalman_Filter_Math.html)

若不建立扎实的数学基础，我们已难以再深入。本章可选，尤其首次阅读时；但若打算编写稳健、数值稳定的滤波器，或阅读文献，则需掌握本章内容。

*本章仍需大量修订。*


[**第 8 章：设计卡尔曼滤波器（Designing Kalman Filters）**](08_Designing_Kalman_Filters.html)

在第 6 章基础上，引导你设计多个卡尔曼滤波器。讨论数值稳定性等问题，但不给出完整解法。


[**第 9 章：非线性滤波（Nonlinear Filtering）**](09_Nonlinear_Filtering.html)

前述卡尔曼滤波仅适用于线性（linear）问题。本章介绍非线性系统（nonlinear systems）给滤波带来的困难，并简要讨论后续章节将学习的、适用于非线性系统的各类算法。


[**第 10 章：无迹卡尔曼滤波（Unscented Kalman Filters）**](10_Unscented_Kalman_Filter.html)

无迹卡尔曼滤波（Unscented Kalman Filter, UKF）是卡尔曼滤波理论的新进展。它可在不要求扩展卡尔曼滤波（Extended Kalman Filter, EKF）那样的闭式解的情况下，对非线性问题进行滤波。


[**第 11 章：扩展卡尔曼滤波（Extended Kalman Filters）**](11_Extended_Kalman_Filters.html)

前述卡尔曼滤波仅适用于线性问题。扩展卡尔曼滤波（Extended Kalman Filter, EKF）是线性化非线性问题最常用的方法。

*本章仍处非常早期的撰写阶段。*


[**第 12 章：设计非线性卡尔曼滤波器（Designing Nonlinear Kalman Filters）**](12_Designing_Nonlinear_Kalman_Filters.html)

通过若干实例讲解如何为非线性问题设计卡尔曼滤波器。*本章仍在大量撰写中。*


[**第 13 章：平滑（Smoothing）**](13_Smoothing.html)

卡尔曼滤波是递推的，因而非常适合实时滤波（real-time filtering）；但它同样非常适用于数据后处理。本章讨论若干常见方法。


[**第 14 章：自适应滤波（Adaptive Filtering）**](14_Adaptive_Filtering.html)
    
卡尔曼滤波通常假设单一过程模型（process model），而机动目标往往需用多种过程模型描述。自适应滤波（adaptive filtering）用多种技术使卡尔曼滤波能适应目标行为的变化。


[**第 15 章：H∞ 滤波（H-Infinity Filters）**](15_HInfinity_Filters.html)
    
介绍 $H_\infty$ 滤波（$H_\infty$ filter）。

*已有实现该滤波器的代码，但尚无配套文字说明。*


[**第 16 章：集总卡尔曼滤波（Ensemble Kalman Filters）**](16_Ensemble_Kalman_Filters.html)

介绍集总卡尔曼滤波（ensemble Kalman filter），它采用蒙特卡洛（Monte Carlo）方法处理非线性系统中规模很大的卡尔曼滤波状态。
