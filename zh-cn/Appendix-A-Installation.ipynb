# 安装（Installation）

本书使用 Jupyter Notebook（基于浏览器的交互式 Python 环境）编写，它将 Python、文本与数学公式融为一体。我选择它是因为其交互特性——在交互式环境中动手实践之前，我几乎无法学会卡尔曼滤波（Kalman filtering）。在能够修改参数并立即看到输出之前，你很难对许多参数形成直觉。交互式环境还允许你进行「如果……会怎样」的设想。「如果我把 $\mathbf{Q}$ 设为零会怎样？」在 Jupyter Notebook 中验证这一点轻而易举。

我选择它的另一个原因是，大多数教科书把许多细节留作不透明。例如，伪代码旁边可能有一张精美的图，但那张图是由软件生成的，而读者却无法使用那款软件。我希望制作本书所用的一切都能供你使用。如何绘制协方差椭圆（covariance ellipse）？读大多数书你不会知道。有了 Jupyter Notebook，你只需查看源代码即可。

即使你选择在线阅读本书，你也需要安装 Python 和 SciPy 套件（SciPy stack），以便编写自己的卡尔曼滤波器（Kalman filters）。安装这些库的方式多种多样，我无法一一涵盖，但会介绍几种典型场景。

## SymPy

SymPy 是用于符号数学（symbolic mathematics）的 Python 包。其全部能力超出本书范围，但它能做代数运算、对方程积分与求导、求微分方程的解，等等。例如，我们用它计算矩阵的雅可比（Jacobian），以及期望值积分。

首先是一个简单示例。我们将导入 SymPy，初始化其漂亮打印功能（pretty print，会用 LaTeX 打印方程）。然后为 SymPy 声明一个符号。

```python
import sympy
sympy.init_printing(use_latex='mathjax')

phi, x = sympy.symbols('\phi, x')
phi
```

注意它如何用 LaTeX 打印符号 `phi`。现在来做点数学：$\sqrt{\phi}$ 的导数是多少？

```python
sympy.diff('sqrt(phi)')
```

我们可以对方程因式分解

```python
sympy.factor(phi**3 -phi**2 + phi - 1)
```

也可以展开它们。

```python
((phi+1)*(phi-4)).expand()
```

你可以为变量的特定取值求方程的值：

```python
w =x**2 -3*x +4
print(w.subs(x, 4))
print(w.subs(x, 12))
```

对于使用你尚未定义符号的方程，你也可以使用字符串：

```python
x = sympy.expand('(t+1)*2')
x
```

现在用 SymPy 计算矩阵的雅可比（Jacobian）。给定函数

$$h=\sqrt{(x^2 + z^2)}$$

求关于 x、y 和 z 的雅可比。

```python
x, y, z = sympy.symbols('x y z')

H = sympy.Matrix([sympy.sqrt(x**2 + z**2)])

state = sympy.Matrix([x, y, z])
H.jacobian(state)
```

现在计算离散过程噪声矩阵（process noise matrix）$\mathbf Q$，已知连续过程噪声矩阵
$$\mathbf Q = \Phi_s \begin{bmatrix}0&0&0\\0&0&0\\0&0&1\end{bmatrix}$$

积分为

$$\mathbf Q = \int_0^{\Delta t} \mathbf F(t)\mathbf Q\mathbf F^T(t)\, dt$$

其中
$$\mathbf F(\Delta t) = \begin{bmatrix}1 & \Delta t & {\Delta t}^2/2 \\ 0 & 1 & \Delta t\\ 0& 0& 1\end{bmatrix}$$

```python
dt = sympy.symbols('\Delta{t}')
F_k = sympy.Matrix([[1, dt, dt**2/2],
                    [0,  1,      dt],
                    [0,  0,       1]])
Q = sympy.Matrix([[0,0,0],
                  [0,0,0],
                  [0,0,1]])

sympy.integrate(F_k*Q*F_k.T,(dt, 0, dt))
```
