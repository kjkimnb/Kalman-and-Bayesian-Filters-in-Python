直观介绍 g-h 滤波（g-h filter），亦称 $\alpha$-$\beta$ 滤波（$\alpha$-$\beta$ filter）。该滤波器族包含卡尔曼滤波（Kalman filter）。读完本章，你将理解卡尔曼滤波背后的核心概念。

[**第 2 章：离散贝叶斯滤波（The Discrete Bayes Filter）**](./02-Discrete-Bayes.ipynb)

介绍离散贝叶斯滤波（discrete Bayes filter）。你将由此学习支撑卡尔曼滤波的、易于理解的概率（贝叶斯，Bayesian）推理。

[**第 3 章：概率、高斯分布与贝叶斯定理（Probabilities, Gaussians, and Bayes' Theorem）**](./03-Gaussians.ipynb)

介绍在贝叶斯（Bayesian）意义下用高斯分布（Gaussian）表示信念。高斯分布使离散贝叶斯滤波中的算法可应用于连续域。

[**第 4 章：一维卡尔曼滤波（One Dimensional Kalman Filters）**](./04-One-Dimensional-Kalman-Filters.ipynb)

通过将离散贝叶斯滤波改为使用高斯分布，实现卡尔曼滤波（Kalman filter）。这是功能完整的一维卡尔曼滤波，但仅适用于一维问题。

[**第 5 章：多元高斯分布（Multivariate Gaussians）**](./05-Multivariate-Gaussians.ipynb)

将高斯分布推广到多维，并说明“三角测量（triangulation）”与隐变量（hidden variables）如何大幅改善估计。

[**第 6 章：多元卡尔曼滤波（Multivariate Kalman Filter）**](./06-Multivariate-Kalman-Filters.ipynb)

将单变量章节的卡尔曼滤波推广为线性问题的完整广义滤波器。读完后你将理解卡尔曼滤波如何工作，以及如何为所选（线性）问题设计与实现滤波器。

[**第 7 章：卡尔曼滤波数学（Kalman Filter Math）**](./07-Kalman-Filter-Math.ipynb)

若不建立扎实的数学基础，我们已难以再深入。本章可选，尤其首次阅读时；但若打算编写稳健、数值稳定的滤波器，或阅读文献，则需掌握本章内容。部分章节为理解后续非线性滤波（nonlinear filtering）所必需。

[**第 8 章：设计卡尔曼滤波器（Designing Kalman Filters）**](./08-Designing-Kalman-Filters.ipynb)

在第 5、6 章基础上，引导你设计多个卡尔曼滤波器。只有看过多种实例，才能真正掌握全部理论。示例力求贴近实际，而非“玩具”问题，以便你开始实现自己的滤波器。讨论数值稳定性等问题，但不给出完整解法。

[**第 9 章：非线性滤波（Nonlinear Filtering）**](./09-Nonlinear-Filtering.ipynb)

前述卡尔曼滤波仅适用于线性（linear）问题，而世界是非线性的。本章介绍非线性系统（nonlinear systems）给滤波带来的困难，并简要讨论后续章节将学习的各类算法。

[**第 10 章：无迹卡尔曼滤波（Unscented Kalman Filters）**](./10-Unscented-Kalman-Filter.ipynb)

无迹卡尔曼滤波（Unscented Kalman Filter, UKF）是卡尔曼滤波理论的新进展。它可在不要求扩展卡尔曼滤波（Extended Kalman Filter, EKF）那样的闭式解的情况下，对非线性问题进行滤波。

现有教材往往略过或一带而过该主题，而把篇幅主要留给 EKF。本书将其提前，因为 UKF 更易理解与实现，滤波性能通常不逊于甚至优于 EKF。面对实际问题，我总会先尝试 UKF，你也应如此。

[**第 11 章：扩展卡尔曼滤波（Extended Kalman Filters）**](./11-Extended-Kalman-Filters.ipynb)

扩展卡尔曼滤波（Extended Kalman Filter, EKF）是线性化非线性问题最常用的方法。现实世界中大多数卡尔曼滤波器都是 EKF，因此需要掌握本章内容，才能理解现有代码、论文与报告等。

[**第 12 章：粒子滤波（Particle Filters）**](./12-Particle-Filters.ipynb)

粒子滤波（particle filter）用蒙特卡洛（Monte Carlo）技术进行滤波。它可轻松处理强非线性、非高斯系统以及多模态分布（multimodal distributions，如同时跟踪多个目标），但计算代价较高。

[**第 13 章：平滑（Smoothing）**](./13-Smoothing.ipynb)

卡尔曼滤波是递推的，因而非常适合实时滤波（real-time filtering）；但它同样非常适用于数据后处理。毕竟卡尔曼滤波是预测-校正（predictor-corrector）结构，预测过去比预测未来容易得多！本章讨论若干常见方法。

[**第 14 章：自适应滤波（Adaptive Filtering）**](./14-Adaptive-Filtering.ipynb)
    
卡尔曼滤波通常假设单一过程模型（process model），而机动目标往往需用多种过程模型描述。自适应滤波（adaptive filtering）用多种技术使卡尔曼滤波能适应目标行为的变化。
